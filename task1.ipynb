```python
import pandas as pd

# Sample dataset for Task 1
data = {
    'Age': [22, 25, 30, 28, 35, 40, 45, 23, 27, 31, 33, 36, 29, 26, 32],
    'Gender': ['Female','Male','Female','Female','Male','Female','Male','Female','Male','Male','Female','Female','Male','Female','Male']
}

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'], bins=8)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x=df['Gender'])
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```
